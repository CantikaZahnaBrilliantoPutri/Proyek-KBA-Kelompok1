Silver Layer Processing - Grocery Inventory Dataset

Notebook ini digunakan untuk proses Data Engineering pada Silver Layer:
- Data Cleaning
- Data Transformation
- Feature Engineering
- Data Validation
- Export ke Silver Layer

```python
import pandas as pd
```

1. Load Data (Bronze Layer)

```python

df = pd.read_csv('Grocery_Inventory new v1.csv')
df.head()
```

2. Data Cleaning

```python
print(df.isnull().sum())

df = df.drop_duplicates()

df = df.dropna()

df.head()
```

3. Data Transformation

```python

df['Unit_Price'] = df['Unit_Price'].str.replace('$', '', regex=False)

df['Unit_Price'] = df['Unit_Price'].astype(float)
```

4. Feature Engineering

```python
df['Inventory_Value'] = df['Stock_Quantity'] * df['Unit_Price']

df['Low_Stock'] = df['Stock_Quantity'] < df['Reorder_Level']

df.head()
```

5. Data Validation

```python
df = df[df['Stock_Quantity'] >= 0]
df = df[df['Unit_Price'] >= 0]

df.describe()
```

6. Save ke Silver Layer

```python
df.to_csv('silver_inventory.csv', index=False)
print('File berhasil disimpan sebagai silver_inventory.csv')
```
